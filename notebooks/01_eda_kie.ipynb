```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/raw/data.csv")

# Basic dataset info
print("Total Customers:", df.shape[0])
print("Total Days:", df.shape[1] - 2)

print("\nTheft vs Normal:")
print(df['FLAG'].value_counts())

# Safe theft percentage calculation
theft_count = df['FLAG'].value_counts().get(1, 0)

print("\nTheft percentage:",
      round(theft_count / len(df) * 100, 2), "%")
```

```python
# Load dataset if not already loaded
if 'df' not in locals():
	df = pd.read_csv('../data/raw/data.csv')

# Check missing values
reading_cols = [col for col in df.columns if col not in ['CONS_NO', 'FLAG']]

missing = df[reading_cols].isnull().sum().sum()
total = df.shape[0] * len(reading_cols)

print(f"Total reading cells: {total:,}")
print(f"Missing values: {missing:,}")
print(f"Missing percentage: {round(missing/total*100, 2)}%")

# Per customer missing rate
df['missing_rate'] = df[reading_cols].isnull().mean(axis=1)
print(f"\nAvg missing rate per customer: {round(df['missing_rate'].mean()*100, 2)}%")
print(f"Max missing rate: {round(df['missing_rate'].max()*100, 2)}%")
```

```python
# Plot class distribution
plt.figure(figsize=(8, 5))
colors = ['#2ecc71', '#e74c3c']
df['FLAG'].value_counts().plot(kind='bar', color=colors)
plt.title('Normal vs Theft Customers', fontsize=14)
plt.xlabel('Class (0=Normal, 1=Theft)')
plt.ylabel('Number of Customers')
plt.xticks([0, 1], ['Normal (91.47%)', 'Theft (8.53%)'], rotation=0)
plt.tight_layout()
plt.savefig('../data/processed/class_distribution.png')
plt.show()
print("Plot saved!")
```

```python
# Compare average consumption: Normal vs Theft
reading_cols = [col for col in df.columns 
                if col not in ['CONS_NO', 'FLAG', 'missing_rate']]

normal_avg = df[df['FLAG']==0][reading_cols].mean(axis=1).mean()
theft_avg  = df[df['FLAG']==1][reading_cols].mean(axis=1).mean()

print(f"Normal customers avg daily usage: {round(normal_avg, 2)} kWh")
print(f"Theft  customers avg daily usage: {round(theft_avg, 2)} kWh")
print(f"Theft uses {round((1 - theft_avg/normal_avg)*100, 1)}% less than normal")
```

```python
# Visualize the difference
plt.figure(figsize=(8, 5))
categories = ['Normal Customers', 'Theft Customers']
values = [normal_avg, theft_avg]
colors = ['#2ecc71', '#e74c3c']

bars = plt.bar(categories, values, color=colors, width=0.4)
plt.title('Average Daily Consumption: Normal vs Theft', fontsize=14)
plt.ylabel('Average Daily Usage (kWh)')

# Add value labels on bars
for bar, val in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, 
             bar.get_height() + 0.3,
             f'{round(val, 2)} kWh', 
             ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig('../data/processed/consumption_comparison.png')
plt.show()
```

```python
# Missing values distribution
plt.figure(figsize=(10, 5))
plt.hist(df[df['FLAG']==0]['missing_rate'], 
         bins=50, alpha=0.6, color='#2ecc71', 
         label='Normal')
plt.hist(df[df['FLAG']==1]['missing_rate'], 
         bins=50, alpha=0.6, color='#e74c3c', 
         label='Theft')
plt.title('Missing Values Distribution: Normal vs Theft', fontsize=14)
plt.xlabel('Missing Rate per Customer')
plt.ylabel('Number of Customers')
plt.legend()
plt.tight_layout()
plt.savefig('../data/processed/missing_distribution.png')
plt.show()
```

```python
# Sample consumption patterns
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Normal customer sample
normal_sample = df[df['FLAG']==0].iloc[0]
reading_cols = [col for col in df.columns 
                if col not in ['CONS_NO', 'FLAG', 'missing_rate']]

axes[0].plot(normal_sample[reading_cols].values, 
             color='#2ecc71', linewidth=0.8)
axes[0].set_title('Normal Customer - Consumption Pattern', fontsize=13)
axes[0].set_ylabel('kWh')
axes[0].set_xlabel('Days')

# Theft customer sample
theft_sample = df[df['FLAG']==1].iloc[0]
axes[1].plot(theft_sample[reading_cols].values, 
             color='#e74c3c', linewidth=0.8)
axes[1].set_title('Theft Customer - Consumption Pattern', fontsize=13)
axes[1].set_ylabel('kWh')
axes[1].set_xlabel('Days')

plt.tight_layout()
plt.savefig('../data/processed/consumption_patterns.png')
plt.show()
```
